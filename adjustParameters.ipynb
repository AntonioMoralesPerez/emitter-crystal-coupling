```python
import numpy as np
from scipy.optimize import least_squares


def fobj(x):
    return np.array(
        [
            100 * (x[0] + 4 * (x[1] + x[2]) - 0) ** 2,  # gamma
            (x[0] - 4 * x[1] + 4 * x[2] - (0.40022819)**2) ** 2,  # m
            (x[0] - 4 * x[2] - (0.27619449)**2) ** 2,  # x
        ]
    )
```

```python
res_1 = least_squares(fobj, np.array([0, 0, 0]))
print(res_1.x)
```
